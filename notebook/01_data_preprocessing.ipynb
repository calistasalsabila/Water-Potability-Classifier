### 01. Data Preprocessing

#### Import the Libraries

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

#### Load the Dataset

```python
df = pd.read_csv('../data/raw/water_potability.csv')
```

#### Inspect the Dataset and Check for Missing Values

```python
# view dataset information
info = df.info
print(info)
```

```python
# check for missing data
missing_data = df.isnull().sum()
print(missing_data)
```

#### Perform EDA: Descriptive Statistics and Visualization

```python
# check descriptive statistics
desc = df.describe()
print(desc)
```

```python
# plot target distrubution
plt.figure(figsize=(6, 4))
sns.countplot(x='Potability', data=df)
plt.title('Distribution of Potability')
plt.show()
```

```python
# plot outliar feature
plt.figure(figsize=(15, 7))
sns.boxplot(data=df[['ph', 'Hardness', 'Chloramines', 'Sulfate', 'Conductivity', 'Organic_carbon', 'Trihalomethanes', 'Turbidity']])
plt.title('Boxplot the Feature')
plt.show()
```

#### Clean the Dataset and Handle Missing Values

```python
df_clean = df.fillna(df.median())

print('remaining missing values after fillna: ')
print(df_clean.isnull().sum())
```

#### Save the Cleaned Dataset

```python
df_clean.to_csv('../data/processed/water_potability_clean.csv', index=False)
print("\nFile saved!")
```
